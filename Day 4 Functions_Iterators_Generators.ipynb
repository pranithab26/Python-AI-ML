## 7. Generators:

#A generator is a special type of iterator.

It is written like a function, but instead of return, it uses yield.

Important idea:

return ends the function.

yield pauses the function.

The generator remembers where it stopped.

It resumes from the same place next time.

It generates values one by one and saves memory.

```python
def generate_cubes(n):
    for number in range(n):
        yield number ** 3

generate_cubes(5)

#When we call a generator directly, it creates a generator object. To get values, we use a for loop.
```

```python
for value in generate_cubes(9):
    print(value)
     
```

```python
#8. Generator Example: Fibonacci Series

#Fibonacci series starts like this:

#1, 1, 2, 3, 5, 8, 13, ...

#Each number is made by adding the previous two numbers.

def generate_fibonacci(n):
    a = 1
    b = 1

    for i in range(n):
        yield a
        a, b = b, a + b


for num in generate_fibonacci(10):
    print(num)
```

```python
#9. Normal Function vs Generator

#A normal function can create and return the full list. A generator gives one value at a time.

def fibonacci_list(n):
    a = 1
    b = 1
    output = []

    for i in range(n):
        output.append(a)
        a, b = b, a + b

    return output

result = fibonacci_list(12)
print(result)

```

```python
def fibonacci_generator(n):
    a = 1
    b = 1

    for i in range(n):
        yield a
        a, b = b, a + b

for value in fibonacci_generator(12):
    print(value)
```

```python
#10. map() Function

#map() applies one function to every item in an iterable.

#Syntax:

#map(function, iterable)

#Example idea:

#[1, 2, 3, 4] mapped with square function gives [1, 4, 9, 16].
```

```python
def square_number(x):
    return x ** 2

numbers = [1, 2, 3, 4, 5]

squared_numbers = list(map(square_number, numbers))

print(squared_numbers)
     
```

```python
#11. Temperature Conversion using map()

#We create conversion functions and apply them to every value in the temperature list.

def fahrenheit(celsius):
    return (float(9) / 5) * celsius + 32

def celsius(fahrenheit_value):
    return (float(5) / 9) * (fahrenheit_value - 32)

temperatures = [0, 22.5, 40, 100]

fahrenheit_values = list(map(fahrenheit, temperatures))
print("Fahrenheit values:", fahrenheit_values)

celsius_values = list(map(celsius, fahrenheit_values))
print("Back to Celsius:", celsius_values)
```

```python
#14. reduce() Function

reduce() reduces many values into one final value.

Example:

[1, 2, 3, 4] becomes 1 + 2 + 3 + 4 = 10.

We need to import it from functools.
```

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5, 6]

total = reduce(lambda a, b: a + b, numbers)

print("Total:", total)
     
```

```python
#15. Reduce Example: Find Maximum Number
```

```python
#We compare two numbers at a time and keep the larger one. Finally, only the maximum value remains.

from functools import reduce

numbers = [47, 49, 42, 55, 56, 13]

maximum_number = reduce(lambda a, b: a if a > b else b, numbers)

print("Maximum number:", maximum_number)
```

```python
#16. filter() Function

#filter() keeps only the values that satisfy a condition.

#Syntax:

#filter(function, iterable)

#The function must return True or False. Only True values are included.

def even_check(number):
    if number % 2 == 0:
        return True
    else:
        return False

numbers = [1, 2, 3, 4, 5, 6, 7, 8]

even_numbers = list(filter(even_check, numbers))

print(even_numbers)
```

```python
#17. Filter with Lambda

#Instead of creating a separate function, we can use lambda.

numbers = [1, 2, 3, 4, 5, 6, 7, 8]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)

```

```python
numbers = [10, 15, 20, 25, 30, 35, 40]

greater_than_25 = list(filter(lambda x: x > 25, numbers))

print(greater_than_25)
```

```python
#18. Filter Example with Skills

names = ["John", "Pranitha", "Sam", "Michael", "rosie"]

long_names = list(filter(lambda name: len(name) >= 5, names))

print(long_names)
```

```python
#19. Difference Between map(), reduce(), and filter()
#Function	Purpose	Output
#map()	Applies a function to every item	Usually returns same number of items
#reduce()	Combines many values into one	Returns one final value
#filter()	Keeps only values that satisfy condition	Returns selected values

#Simple idea:

#map() = transform every value
#reduce() = combine values into one
#filter() = select matching values

```

```python
from functools import reduce

prices = [100, 200, 300, 400, 500]

double_prices = list(map(lambda price: price * 2, prices))
high_prices = list(filter(lambda price: price >= 300, prices))
total_price = reduce(lambda a, b: a + b, prices)

print("Original:", prices)
print("Double prices using map:", double_prices)
print("Prices >= 300 using filter:", high_prices)
print("Total using reduce:", total_price)
```

```python
#20. Practice Example: Pranitha Hobbies
#This example combines:

#list
#function
#lambda
#map
#filter
#reduce
```

```python
from functools import reduce

courses = ["Python", "SQL", "Power BI", "Tableau", "Cloud"]

upper_courses = list(map(lambda course: course.upper(), courses))
long_courses = list(filter(lambda course: len(course) > 5, courses))
all_courses = reduce(lambda a, b: a + ", " + b, courses)

print("Original courses:", courses)
print("Uppercase courses:", upper_courses)
print("Long courses:", long_courses)
print("Combined courses:", all_courses)
     
```
