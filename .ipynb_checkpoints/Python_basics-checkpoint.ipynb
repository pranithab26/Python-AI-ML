## 3. Functions

Functions in Python are used in situations which requires using a specific code block multiple times.

For example, If we need to calculate the area of 10 squares with different side lengths, we can define a function which takes the side length as an argument and returns the area of the square. We can then use ths function directly instead of writing the entire code for calculating area everytime.

```python
def area_of_square(side):
    return side*side
squares = [2,5,6,8,9,10,7,1,3,4]
for i in squares:
    print(area_of_square(i))
```
