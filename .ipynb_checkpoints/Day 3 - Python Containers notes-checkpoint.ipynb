#1. What are containers in Python?

#A container is a data type that can store multiple values.

#In Day 3, the main containers are:

#List: ordered, mutable, allows duplicates
#Tuple: ordered, immutable, allows duplicates
#Set: unordered, mutable, stores only unique values
#Dictionary:ordered,mutable, key-value pair based container(keys must be unique).
#Containers help us store student details, skills, marks, project information, and application data in an organized way.

```python
name = "Pranitha"

courses = ["Python", "SQL", "Power BI", "Tableau"]

education = ("MS in Computer and Information Systems", "Rivier University", 2023)

favorite_numbers = {10, 20, 10, 30}

student = {
    "name": "Pranitha",
    "university": "Rivier University",
    "degree": "MS CIS"
}

print("String:", name)
print("List:", courses)
print("Tuple:", education)
print("Set removes duplicates:", favorite_numbers)
print("Dictionary:", student)
```

```python
##2. Lists
#A list stores multiple values inside square brackets [].

#Important points:
#Lists are ordered.
#Lists allow duplicate values.
#Lists can store mixed data types.
#Lists are mutable, which means we can modify them after creation.
```

```python
# Creating lists
favorite_foods = ["Pizza", "Pasta", "Ice Cream", "Burger"]
person_info = ["Pranitha", 25, "Boston", "Student", True]

print("Favorite foods:", favorite_foods)
print("Person information:", person_info)
print("Type of favorite_foods:", type(favorite_foods))
```

```python
#3. List indexing and slicing

#Python indexing starts from 0.
#For a list:

favorite_foods = ["Pizza", "Pasta", "Ice Cream", "Burger"]
#index 0 gives the first item
#index 1 gives the second item
#negative index -1 gives the last item
#Slicing is used to extract part of a list.
```

```python
favorite_foods = ["Pizza", "Pasta", "Ice Cream", "Burger"]

print("Full list:",favorite_foods )
print("Index 0:", favorite_foods[0])
print("Index 2:", favorite_foods[2])
print("Last item:", favorite_foods[-1])
print("First three skills:", favorite_foods[0:3])
print("Every second skill:", favorite_foods[0:5:2])
```

```python
#4. List mutability
#Lists are mutable. This means we can change values after the list is created.
```

```python
favorite_foods = ["Pizza", "Pasta", "Ice Cream", "Burger"]
print("before change",favorite_foods)

favorite_foods[1]="sub"
print("After change",favorite_foods)
```

```python
#5. Basic list methods
#Common list methods:

#append() adds one item at the end
#extend() adds multiple items separately
#insert() adds an item at a specific index
#pop() removes an item and can save it
#remove() removes a specific value
#reverse() reverses the list permanently
#sort() sorts the list permanently
```

```python
fruits = ["Apple", "Banana", "Orange"]
print("Original:", fruits)

fruits.append("Mango")
print("After append:", fruits)

fruits.extend(["Grapes", "Pineapple"])
print("After extend:", fruits)

fruits.insert(1, "Watermelon")
print("After insert:", fruits)

removed_fruit = fruits.pop()
print("Popped item:", removed_fruit)
print("After pop:", fruits)

fruits.remove("Orange")
print("After remove Orange:", fruits)

fruits.reverse()
print("After reverse:", fruits)

fruits.sort()
print("After sort:", fruits)
```

```python
#6. Difference between append and extend
#append() adds the complete object as one item.

#extend() breaks the incoming list and adds items one by one.
```

```python
fruits_append = ["Apple", "Banana"]
fruits_append.append(["Orange", "Mango"])
print("Using append:", fruits_append)

fruits_extend = ["Apple", "Banana"]
fruits_extend.extend(["Orange", "Mango"])
print("Using extend:", fruits_extend)
```

```python
#7. Nested lists
#A nested list means a list inside another list.

#This is useful for matrix-like data or grouped data.

#Example: storing three semesters of courses.
```

```python
team_1 = ["John", "Emma", "David"]
team_2 = ["Sophia", "Michael", "James"]
team_3 = ["Olivia", "Daniel", "Ethan"]

company_teams = [team_1, team_2, team_3]

print("Company teams:", company_teams)
print("First team:", company_teams[0])
print("Second team first employee:", company_teams[1][0])
print("Third team third employee:", company_teams[2][2])
```

```python
#8. List comprehension
#List comprehension is a shorter way to create a list using a loop.

#Normal way:

#List Comprehension

#List comprehension is a concise (shorter) way to create a new list from an existing iterable (such as a list, tuple, or range) using a single line of code.

#Traditional Method (Using a Loop)

numbers = [1, 2, 3, 4, 5]

squares = []

for num in numbers:
    squares.append(num * num)

print(squares)   

## Using List Comprehension
numbers = [1, 2, 3, 4, 5]

squares = [num * num for num in numbers]

print(squares)
```

```python
#9. Tuple
#A tuple stores multiple values inside parentheses ().

#Important points:

#Tuples are ordered.
#Tuples allow duplicates.
#Tuples are immutable, which means we cannot change them directly after creation.
#Use tuples when data should not be changed, such as student ID, university name, or graduation year.
```

```python
employee_record = ("Pranitha", "Salesforce Developer", "Zscaler", 3)

print("Employee record:", employee_record)
print("Employee Name:", employee_record[0])
print("Job Title:", employee_record[1])
print("Type:", type(employee_record))
```

```python
#10. Tuple immutability
#A tuple cannot be changed directly. If we try to change a tuple item, Python gives an error.

#To avoid stopping the notebook, the example below uses try/except.
```

```python
employee_record = ("Pranitha", "Salesforce Developer", "Zscaler", 3)

try:
    employee_record[2] = "Google"
except TypeError as error:
    print("Error because tuple is immutable:", error)
```

```python
#11. How to update tuple data indirectly
#If we really need to update a tuple:

#Convert tuple to list
#Modify the list
#Convert list back to tuple
```

```python
car_info = ("Ford", "Mustang", 2025)

car_list = list(car_info)
car_list[1] = "Explorer"
car_info = tuple(car_list)

print("Updated tuple:", car_info)
```

```python
#12. Sets
#A set stores unique values inside curly braces {}.

#Important points:

#Sets remove duplicate values.
#Sets are unordered.
#Sets are useful when we only want unique data.
#Example: unique skills from repeated values.
```

```python
sports = {"Cricket", "Football", "Cricket", "Tennis", "Football", "Basketball"}

print("Unique sports:", sports)
print("Type:", type(sports))
```

```python
#13. Set methods
#Common set methods:

#add() adds one item
#remove() removes item but gives error if item is missing
#discard() removes item safely even if missing
#clear() removes everything
#pop() removes one item from the set
```

```python
fruits = {"Apple", "Banana", "Mango"}
print("Original set:", fruits)

fruits.add("Orange")
print("After add:", fruits)

fruits.discard("Grapes")
print("After discard missing value Grapes:", fruits)

try:
    fruits.remove("Grapes")
except KeyError as error:
    print("remove() error because Grapes is missing:", error)

popped_value = fruits.pop()
print("Popped value:", popped_value)
print("After pop:", fruits)

fruits.clear()
print("After clear:", fruits)
```

```python
#14. Convert list to set to remove duplicates
#If a list has duplicate values, convert it to a set to keep only unique values.

languages_list = ["Java", "Python", "Java", "C++", "Python", "Go", "C++"]

unique_languages_set = set(languages_list)
unique_languages_list = list(unique_languages_set)

print("Original list:", languages_list)
print("Unique set:", unique_languages_set)
print("Back to list:", unique_languages_list)
```

```python
#15. Dictionaries
#A dictionary stores data as key-value pairs.

#Syntax:

my_dict = {"key": "value"}
#Important points:

## Dictionary uses keys instead of index positions.
#Keys should be unique.
#Values can be strings, numbers, lists, tuples, sets, or even another dictionary.
#Dictionaries are mutable.
```

```python
employee = {
    "name": "Smith",
    "company": "TechCorp",
    "role": "Software Engineer",
    "experience_years": 5,
    "skills": ["Java", "Spring Boot", "AWS"]
}

print("Employee dictionary:", employee)
print("Name:", employee["name"])
print("Skills:", employee["skills"])
print("First skill:", employee["skills"][0])
```

```python
#16. Updating dictionary values
#Dictionaries are mutable, so we can add or update key-value pairs.

employee = {
    "name": "Smith",
    "company": "TechCorp",
    "role": "Software Engineer"
}

print("Before update:", employee)

employee["company"] = "Google"
employee["location"] = "California"
employee["specialization"] = "Cloud Computing"

print("After update:", employee)
```

```python
#17. Dictionary keys, values, and items
#Useful dictionary methods:

#keys() gives all keys
#values() gives all values
#items() gives key-value pairs

product = {
    "name": "Laptop",
    "brand": "Dell",
    "price": 85000,
    "year": 2025
}

print("Keys:", product.keys())
print("Values:", product.values())
print("Items:", product.items())

print("\nReading using loop:")
for key, value in product.items():
    print(key, "-->", value)
```

```python
#18. Duplicate keys in dictionary
#Dictionary keys must be unique.

#If the same key is repeated, Python keeps the latest value.

product = {
    "item": "Laptop",
    "price": 50000,
    "price": 65000,
    "brand": "Dell"
}

print(product)
print("Python kept the latest value for the duplicate key 'price'.")
```

```python
#19. Nested dictionary
#A nested dictionary means a dictionary inside another dictionary.

#This is useful for storing structured data like profile, education, and skills.


employee_profile = {
    "personal": {
        "name": "Smith",
        "city": "Austin"
    },
    "work": {
        "company": "TechCorp",
        "role": "Software Engineer",
        "experience_years": 5
    },
    "skills": {
        "programming": ["Java", "Python", "Go"],
        "tools": ["AWS", "Docker", "Kubernetes"]
    }
}

print("Full profile:", employee_profile)
print("Name:", employee_profile["personal"]["name"])
print("Role:", employee_profile["work"]["role"])
print("First programming skill:", employee_profile["skills"]["programming"][0])
```

```python
#20. Dictionary comprehension
#Dictionary comprehension is a short way to create dictionaries using loops.

#Example: create a dictionary where each number is mapped to its square.

numbers = [1, 2, 3, 4, 5]

squares = {num: num ** 2 for num in numbers}

print("Numbers:", numbers)
print("Squares dictionary:", squares)
```

```python
#21. Basic functions
#A function is a reusable block of code.

#Syntax:

def function_name():
    code
#Functions help us organize code and avoid writing the same logic again and again.
```

```python
def show_employee_details():
    name = "Smith"
    company = "TechCorp"
    role = "Software Engineer"

    print("Name:", name)
    print("Company:", company)
    print("Role:", role)

show_employee_details()
```

```python
#22. Function with parameters and return value
#A function can take input using parameters and send output using return.

def calculate_average(marks):
    total = sum(marks)
    average = total / len(marks)
    return average

my_marks = [70, 80, 90]
result = calculate_average(my_marks)

print("Marks:", my_marks)
print("Average:", result)
```

```python
#23. Small container summary function

#This example combines lists, tuples, sets, dictionaries, and functions.

def summarize_employee():
    skills_list = ["Java", "AWS", "Java", "Docker"]
    experience_tuple = ("Software Engineer", "TechCorp", 5)
    unique_skills = set(skills_list)

    employee_profile = {
        "name": "John Smith",
        "experience": experience_tuple,
        "unique_skills": list(unique_skills)
    }

    print("List with duplicates:", skills_list)
    print("Tuple experience info:", experience_tuple)
    print("Set unique skills:", unique_skills)
    print("Dictionary profile:", employee_profile)

summarize_employee()
```
