## What is __str__() in Python?
__str__() is a special (magic/dunder) method used to define how an object should be represented as a string when you use:
.print(object)
.str(object)

Without __str__(), Python displays a default object reference that is not very user-friendly.The __str__() method is used to define how an object should be displayed when you print it.

```python
##Example Without __str__()

class Student:
    def __init__(self, name):
        self.name = name

s1 = Student("John")

print(s1)

print(s1)   ##This output is not very meaningful.
```

```python
##Example With __str__()
##Now the object is displayed in a readable format.

class Student:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return "Student Name: " + self.name

s1 = Student("John")

print(s1)
```

```python
##__repr__() is another special method in Python.

##What is __repr__()?
##__repr__() is a special method that defines the official string representation of an object. It is called by repr() and is mainly used for debugging and development purposes.
```

```python
## with __ repr__

class Student:
    def __init__(self, name):
        self.name = name

    def __repr__(self):
        return "Student('" + self.name + "')"

s1 = Student("John")

print(repr(s1))

##How it works
##repr(p1) calls the __repr__() method.
##__repr__() returns the string "Person('John')".
```

```python
## Without __repr__()
class Person:
    def __init__(self, name):
        self.name = name

p1 = Person("John")

print(repr(p1))     ##Since __repr__() is not defined, Python shows the default object representation (memory address).
```

## Abstraction:Hiding unnecessary details
In simple words, the user knows what an object does, but not how it does it.
Abstraction is the process of hiding implementation details and showing only essential features to the user.

## Real-Life Example
You know how to start and drive the car.
You don't need to know how the engine works internally.

This is abstraction.






```python
#### Simple Example

from abc import ABC, abstractmethod

class Remote(ABC):

    @abstractmethod
    def turn_on(self):
        pass

class TV(Remote):

    def turn_on(self):
        print("TV is ON")

tv = TV()
tv.turn_on()
```

## Inheritance:
Inheritance is an OOP (Object-Oriented Programming) concept where one class acquires the properties and methods of another class.

The existing class is called the Parent Class (or Base Class).

The new class is called the Child Class (or Derived Class,sub class).
## Why Use Inheritance?
Reuse code.

Avoid writing the same code multiple times.

Make programs easier to maintain.

```python
## Simple Example

class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    pass

d = Dog()
d.sound()
```

## What are *args and **kwargs in Python?

*args is used to pass multiple positional arguments to a function as a tuple, while **kwargs is used to pass multiple keyword arguments as a dictionary.

They are used to pass multiple arguments to a function.
1. *args (Non-keyword arguments)
Used to pass any number of values
Stored as a tuple

2. **kwargs (Keyword arguments)
Used to pass key-value pairs
Stored as a dictionary


```python
##*args (Non-keyword arguments)
## example 

def add(*args):
    total = 0
    for num in args:
        total += num
    print(total)

add(10, 20, 30)
add(5, 15)

##Explanation
##You can pass any number of values.
##*args collects them like a tuple:
(10, 20, 30)
(5, 15)
```

```python
##**kwargs (Keyword arguments)
## example 

def student_info(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

student_info(name="John", age=20, city="Boston")

##Explanation
##You pass data in key=value format
## Python stores it as a dictionary:
```

```python
##*args → Many values → like a list/tuple
##**kwargs → Many key-value pairs → like a dictionary
```

## Method overriding:
Method overriding happens when a child class provides its own implementation of a method that is already defined in the parent class.

In simple words:Same method name in parent and child class, but different behavior in child class.


```python
## Example:

class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

a = Animal()
a.sound()

d = Dog()
d.sound()
```

## What is Dynamic Extension in Python?
Dynamic Extension is a feature of Python that allows attributes and methods to be added to objects or classes during runtime. This provides flexibility and makes Python a highly dynamic language.

```python
## Example1

class Student:
    pass

s1 = Student()

s1.name = "Pranitha"

print(s1.name)
```

```python
##Example 2

class Employee:
    pass

emp = Employee()

emp.name = "Pranitha"
emp.role = "Developer"

print(emp.name)
print(emp.role)

##Explanation:The Employee class has no attributes defined, but we dynamically added name and role while the program was running.
```

```python
##Example3

class Student:
    pass

def show():
    print("I am a Student")

s1 = Student()

s1.show = show

s1.show()

##Explanation:We created a function show() and attached it to the object s1 at runtime.
```

## Polymorphism in Python
Definition:Polymorphism means "many forms."
The same method can behave differently depending on the object calling it.

```python
## example:

class Dog:
    def sound(self):
        print("Bow Bow")

class Cat:
    def sound(self):
        print("Meow Meow")

dog = Dog()
cat = Cat()

dog.sound()
cat.sound()

## Both classes have a method called sound().

## The same method name behaves differently for different objects.

## This is called Polymorphism.
```

## Duck Typing in Python
##Definition:Duck Typing means:

"If it looks like a duck and quacks like a duck, then treat it as a duck."

Python does not care about the object's class.

It only checks whether the object has the required method.

```python
## Example:

class Duck:
    def fly(self):
        print("Duck is flying")

class Airplane:
    def fly(self):
        print("Airplane is flying")

def start_flying(obj):
    obj.fly()

start_flying(Duck())
start_flying(Airplane())

##Explanation:

##The function:

def start_flying(obj):
    obj.fly()

##doesn't check whether obj is a Duck or Airplane.

##It only expects the object to have a fly() method.

##This is Duck Typing.

##Polymorphism → Same method, different behavior.
##Duck Typing → Same behavior, type doesn't matter.
```

## Class Variables and Instance Variables in Python:

In Python, variables inside a class are mainly of two types:

Class Variables
Instance Variables

1. Class Variables

A class variable is shared by all objects of the class.

It is declared inside the class but outside any method.

```python
##Example:

class Employee:
    company = "Google"     # Class Variable

emp1 = Employee()
emp2 = Employee()

print(emp1.company)
print(emp2.company)

##Explanation:
##company belongs to the class, not to individual objects.
##All objects share the same value.
##Memory is allocated only once.
```

```python
##Changing a Class Variable:

class Employee:
    company = "Google"

emp1 = Employee()
emp2 = Employee()

Employee.company = "Microsoft"

print(emp1.company)
print(emp2.company)   ##Since it's a class variable, the change affects all objects.
```

## Instance Variables:

An instance variable belongs to a specific object.

It is usually created inside the __init__() method using self.

```python
##Example:

class Employee:
    def __init__(self, name):
        self.name = name              # Instance Variable

emp1 = Employee("Pranitha")
emp2 = Employee("John")

print(emp1.name)
print(emp2.name)

## Explanation:
#Each object gets its own copy of name.
#Changing one object's value does not affect another object.
```

```python
## Class Variable + Instance Variable Together
class Employee:
    company = "Google"             # Class Variable

    def __init__(self, name):
        self.name = name           # Instance Variable

emp1 = Employee("Pranitha")
emp2 = Employee("John")

print(emp1.company)
print(emp1.name)

print(emp2.company)
print(emp2.name)

##company is shared.
## name is unique for each object.
```

```python
##Example:

class Student:
    school = "ABC School"    # Class Variable

    def __init__(self, name):
        self.name = name     # Instance Variable

s1 = Student("Pranitha")
s2 = Student("David")

print(s1.school)
print(s2.school)

print(s1.name)
print(s2.name)
```

#Easy way to remember:

#Class Variable → Shared by all students (same school).
#Instance Variable → Unique to each student (different names).

## Instance method,class method,static method:

| Method Type     | Uses                                             | First Parameter    |
| --------------- | ------------------------------------------------ | ------------------ |
| Instance Method | Works with object data                           | `self`             |
| Class Method    | Works with class data                            | `cls`              |
| Static Method   | Utility method, doesn't use object or class data | No `self` or `cls` |


```python
## Instance Method

##An instance method works with the data of a specific object. It uses self as the first parameter.

class Student:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Student Name:", self.name)

s1 = Student("Pranitha")
s1.display()

##Explanation
##display() is an instance method.
##self.name accesses the object's data.
##Different objects can have different values.
    
```

```python
## Class Method

##A class method works with class variables and uses cls as the first parameter.

class Student:
    school = "ABC School"

    @classmethod
    def show_school(cls):
        print("School Name:", cls.school)
        
Student.show_school()

##Explanation:
##@classmethod makes the method a class method.
##cls refers to the class itself.
##It can access class variables.
```

```python
## Static Method

## A static method does not use self or cls.

## It behaves like a normal function placed inside a class.

class Calculator:

    @staticmethod
    def add(a, b):
        print(a + b)

Calculator.add(10, 20)

```

```python
##All three together:

class Student:

    school = "ABC School"      # Class Variable

    def __init__(self, name):
        self.name = name       # Instance Variable

    # Instance Method
    def display(self):
        print("Name:", self.name)

    # Class Method
    @classmethod
    def show_school(cls):
        print("School:", cls.school)

    # Static Method
    @staticmethod
    def greet():
        print("Welcome Students")

s1 = Student("Pranitha")

s1.display()              # Instance Method
Student.show_school()     # Class Method
Student.greet()           # Static Method
```

## A decorator is a function that adds extra functionality to another function without modifying its original code.

#Think of it like adding toppings to a pizza:

##Pizza = Original function
##Toppings = Extra functionality added by the decorator

```python
##Simple Example Without Decorator

def greet():
    print("Hello")

greet()
```

```python
##Simple Example With Decorator

def my_decorator(func):
    def wrapper():
        print("Before function call")
        func()
        print("After function call")
    return wrapper

@my_decorator
def greet():
    print("Hello")

greet()
```

```python
##Example 

def log_decorator(func):
    def wrapper():
        print("Function is starting...")
        func()
        print("Function completed.")
    return wrapper

@log_decorator
def display():
    print("Welcome to Python")

display()
```

```python
##Data class:A data class in Python is a special kind of class that is mainly used to store data (like a container), without needing to write a lot of extra code.

from dataclasses import dataclass

@dataclass
class Student:
    name: str
    age: int
    marks: float

s1 = Student("Sam", 22, 89.5)
s2 = Student("John", 23, 91.0)

print(s1)

##Now we can create objects easily
```

## SRP:SRP (Single Responsibility Principle) in Python is a design principle that says:

A class should have only one reason to change — meaning it should do only one job or responsibility.
means a class should have only one responsibility and only one reason to change.
Why SRP is important?

✔ Easier to understand code
✔ Easier to maintain
✔ Easier to test
✔ Changes don’t break unrelated parts

```python
## Example of SRP violation (bad design)

class Student:
    def __init__(self, name):
        self.name = name

    def save_to_database(self):
        print("Saving student to database")

    def send_email(self):
        print("Sending email to student")

##Explanation:

##This class is doing 3 jobs:

##Handling student data
##Database operations
##Email sending

## If email logic changes, this class must change
##If database logic changes, this class must change
```

```python
##Correct SRP example (good design)

class Student:
    def __init__(self, name):
        self.name = name


class StudentRepository:
    def save(self, student):
        print("Saving student to database")


class EmailService:
    def send_email(self, student):
        print("Sending email to student")

##Explanation: Now each class has ONE responsibility:
##Student → stores student data
##StudentRepository → handles database operations
##EmailService → handles email sending
```
