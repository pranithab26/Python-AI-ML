##Implicit and Explicit Type Conversion (Type Casting) in Python

.Type conversion means changing one data type into another data type.

.There are two types of type conversion in Python:

1.Implicit Type Conversion
2.Explicit Type Conversion

```python
## Example for Implicit type 

a=10
b=2.5
c=a+b
print("The value of c is",c)
print(type(c))                 # type() is used to check the datatype of the variable.

## Explanation:
#a is an integer (int)
#b is a float (float)
#Python automatically converts a to 10.0
#Result becomes 12.5

#This automatic conversion is called Implicit Type Conversion.
#In the above example we can see the c variable type is float. It is because Python always converts smaller data types to larger data types to avoid the loss of data. This is called as Implicit type conversion.
```

```python
##2. Explicit Type Conversion (Type Casting)
## Explicit Type Conversion means the programmer manually converts one data type into another using functions like:

#int()
#float()
#str()
#list()
#tuple()
#set()
```

```python
str_2 = "16"
num_1 = 45

# print("Addition is ", str_2 + num_1)  #we'll get a typeerror in this case, as Python is not able to use Implicit Conversion in such conditions. Thus in this case we need to use Explicit Type conversion.

str_2 = int(str_2)
print("Addition is ", str_2 + num_1)
```

```python
## BRANCHING (if/ elif/ else): 
#example on it if statements 
#Nested if statements.

# Code 1: Write a code to check if the number is even or odd?

num = int(input("Please enter any number you wanted to check: "))

if num % 2 == 0:
    print(f"Entered number {num} is an Even number")
else:
    print(f"Entered number {num} is an Odd number")
```

```python
# Code 3: Understanding of nested if-else statements:

print(" Simple nested if else statement example")

number1 = int(input("Please enter value for number1 "))

if number1 > 10:
    print("number 1 is greater than 10")
    
    if number1 <= 20:
        print("number 1 is between 10 and 20")
    
    elif number1 <= 50:
        print("number 1 is between 10 and 50")
    
    else:
        print("number 1 is greater than 10, 20, 50")
    
else:
    print("number 1 is less than 10 or equal to 10")
```
