#2. File Modes in Python
Mode	Meaning
w	Write mode. Creates a file or overwrites existing content
r	Read mode. Reads existing content
a	Append mode. Adds content at the end without deleting old content
r+	Read and write mode
w+	Write and read mode, but overwrites old content
a+	Append and read mode
Important:

w overwrites old content.
a keeps old content and adds new content.
r only reads content.

#3. Write Data into a Text File

#We use open() with w mode to write data into a file.

#The \n is used to move to the next line.

```python
file_name="employee_details.txt"

with open(file_name,"w")as file:
    file.write("name:Pranitha\n")
    file.write("Position: Python Developer\n")
    file.write("company:Zscaler\n")

    print("file written sucessfully:",file_name)

    
```

```python
#4. Read the Full File
#read() reads the complete file content.

with open("employee_details.txt","r")as file:
    content=file.read()

    print(content)

```

```python
#5. Read File Line by Line

#A for loop can read one line at a time.

#strip() removes extra spaces and new line characters from the left and right sides.
```

```python
with open("employee_details.txt","r") as file:
    for line in file:
        print(line.strip())
```

```python
#6. Write Mode Overwrites Old Content

#If we open the same file again using w, old data will be removed.

with open("employee_details.txt", "w") as file:
    file.write("This file was overwritten using write mode.\n")

with open("employee_details.txt", "r") as file:
    print(file.read())
    
```

```python
#7. Append Mode
# Append mode adds new content without deleting previous content.
```

```python
with open("employee_details.txt","a") as file:
    file.write("new line added using append mode\n")
    file.write("pranitha is learing python with AI-ML")
with open("employee_details.txt","r") as file:
    print(file.read())
```

```python
#8. Read Specific Number of Characters
# read(n) reads only the first n characters.

with open("employee_details.txt","r")as file:
    first_5_char=file.read(5)

print(first_5_char)   
    
```

```python
#9. Read One Line at a Time Using readline()
#Every time we call readline(), the cursor moves to the next line.

with open("employee_details.txt","r") as file:
    line1=file.readline().strip()
    line2=file.readline().strip()
    line3=file.readline().strip()
print("line1", line1)
print("line2", line2)
print("line3", line3)


    
    
```

```python
#10. Read All Lines Using readlines()

#readlines() stores all lines inside a list.

with open("employee_details.txt", "r") as file:
    lines = file.readlines()

print(lines)
print("Total lines:", len(lines))
```

```python
#11. Copy Data from One File to Another

#This is similar to a simple ETL pipeline.

#Read from source file
#Clean each line
#Write to destination file

source_file = "employee_details.txt"
destination_file = "employee_details_copy.txt"

with open(source_file, "r") as src:
    with open(destination_file, "w") as dest:
        for line in src:
            cleaned_line = line.strip()
            dest.write(cleaned_line + "\n")

print("File copied successfully")

with open(destination_file, "r") as file:
    print(file.read())

```

```python
#12. Using seek() with r+
#seek() moves the file cursor to a specific position.

#r+ mode allows reading and writing.

#Be careful because writing at a position can overwrite existing text.

file_name="seek_example.txt"

with open(file_name,"w")as file:
    file.write("Pranitha is learning python")
with open(file_name,"r+") as file:
    file.seek(8)
    file.write("Bantu")
with open(file_name,"r") as file:
    print(file.read())
```

```python
#23. Creating a Custom Module

#Here we create a custom file called mymath_Pranitha.py.

module_code = '''
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b
'''

with open("mymath_pranitha.py", "w") as file:
    file.write(module_code)

print("Custom module created: mymath_pranitha.py")
```

```python
#24. Importing Custom Module

import mymath_Pranitha as mymath

print("Addition:", mymath.add(10, 5))
print("Subtraction:", mymath.subtract(10, 5))
print("Multiplication:", mymath.multiply(10, 5))
print("Division:", mymath.divide(10, 5))
print("Division by zero:", mymath.divide(10, 5))
```

```python
#25. Exception Handling

#Errors can happen in real applications.

#Examples:

#File not found
#Wrong input type
#Division by zero
#Database error
#Missing library
#Permission error
#If errors are not handled, the full application can crash.

#Python uses:

try:
    code that may fail
except:
    code to handle error
finally:
    code that always runs
```

```python
#26. Basic Try-Except

try:
    result = 10 / 0
    print(result)
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
```

```python
#27. Handling Wrong Input
user_input = "rishi"

try:
    number = int(user_input)
    print("Number:", number)
except ValueError:
    print("Error: Please enter only integer values.")
```

```python
#28. Try-Except-Finally

try:
    age = int("26")
    print("Age:", age)
except ValueError:
    print("Invalid age entered.")
finally:
    print("Finally block executed.")
```

```python
#29. File Error Handling

try:
    with open("file_that_does_not_exist.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("Error: File not found. Please check the file path.")
finally:
    print("File operation completed.")
```

#30. SQLite Database Connectivity
#SQLite is a local database.

#Python has a built-in library called sqlite3.

#Steps:

#Connect to database
#Create cursor
#reate table
#Insert data
#Commit
#Query data
#Close connection

```python
import sqlite3

db_file = "Pranitha_database.db"

connection = sqlite3.connect(db_file)
cursor = connection.cursor()

cursor.execute("DROP TABLE IF EXISTS students")

cursor.execute('''
CREATE TABLE students (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    skill TEXT,
    score INTEGER
)
''')

students_data = [
    ("Anil", "Python", 95),
    ("Haran", "Machine Learning", 90),
    ("Rishi", "Linux", 88),
    ("Pranith", "Cybersecurity", 85)
]

cursor.executemany("INSERT INTO students (name, skill, score) VALUES (?, ?, ?)", students_data)

connection.commit()

print("Database created and data inserted:", db_file)
```

```python
#31. Query Data from SQLite

cursor.execute("SELECT * FROM students")
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
#32. Query with Condition

cursor.execute("SELECT name, skill, score FROM students WHERE score >= 90")
high_scores = cursor.fetchall()

for row in high_scores:
    print(row)
```

```python
#33. Order Data Using SQL
cursor.execute("SELECT name, skill, score FROM students ORDER BY score DESC")
ordered_rows = cursor.fetchall()

for row in ordered_rows:
    print(row)

connection.close()
print("Database connection closed.")
```
