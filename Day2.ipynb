```python
## Day 2 - Python Loops and Strings Assignment

#Topic: Assignment Practice - Loops and Strings

#This notebook is separate from notes. It contains only assignment-style problems and runnable solutions.

#The examples use my details to keep the practice simple and easy to understand.

#Assignment 1: Reverse a Single String Without Built-In Reverse

# Problem: Reverse the string Pranitha using a for loop.

text = "Pranitha"
reversed_text = ""

for character in text:
    reversed_text = character + reversed_text

print("Original string:", text)
print("Reversed string:", reversed_text)
```

```python
## Assignment 2: Reverse Full Name Character by Character
##Problem: Reverse Nikhilesh Verma completely.

Full_name= "Pranitha Bantu"
reversed_name=""

for char in Full_name:
    reversed_name=char+reversed_name
    
print("The full name",Full_name)
print("The reversed full name", reversed_name)
```

```python
## Assignment 3: Reverse Each Word but Keep Word Order Same
## Problem: Convert:Pranitha Bantu into ahtinarP utnaB
## Here, each word is reversed, but the word order remains the same.

full_name = "Pranitha Bantu"
words = full_name.split()
result = ""

for word in words:
    reversed_word = ""
    for character in word:
        reversed_word = character + reversed_word
    result += reversed_word + " "

print("Original:", full_name)
print("Each word reversed, same order:", result.strip())
```

```python
## Assignment 4: Reverse Word Order Only
## Problem: Convert Pranitha likes pizza into:pizza likes Pranitha
# Here, the spelling of each word remains same, but the word order changes.

sentence = "Pranitha likes pizza"
words = sentence.split()
reversed_order = ""

for index in range(len(words) - 1, -1, -1):
    reversed_order += words[index] + " "

print("Original sentence:", sentence)
print("Reverse word order:", reversed_order.strip())
```

```python
## Assignment 5: Reverse Each Word and Also Reverse Word Order
# Problem: Convert:Pranitha likes pizza into reversed word order and reversed characters.

sentence = "Pranitha likes pizza"
words = sentence.split()
final_result = ""

for index in range(len(words) - 1, -1, -1):
    word = words[index]
    reversed_word = ""

    for character in word:
        reversed_word = character + reversed_word

    final_result += reversed_word + " "

print("Original sentence:", sentence)
print("Reverse words + reverse order:", final_result.strip())

```

```python
## Assignment 6: Sum of First N Natural Numbers Using While Loop
#Problem: Find the sum of numbers from 1 to 5 using a while loop.

n=5
counter=1
sum_val=0

while counter<n:
    sum_val+=counter
    counter+=1

print("The sum of 1 to 5 numbers is",sum_val)  

```

```python
##Assignment 7: Print Even Numbers Using While Loop
## Problem: Print even numbers from 1 to 10.

number = 1

while number <= 20:
    if number % 2 == 0:
        print(number)
    number += 1 
```

```python
## Assignment 8: Use Break Statement

##Problem: Loop through my full name and stop when character t is found.

name = "Pranitha"

for character in name:
    if character == "t":
        print("Found t. Breaking the loop.")
        break
    print("character")

```

```python
## Assignment 9: Use Continue Statement

## Problem: Print my name but skip the letter i.

name = "Pranitha"

for character in name:
    if character == "i":
        continue
    print("character")
```
