1. For Loop
A for loop is used when we already have an iterable object and we want to go through each item one by one.

Examples of iterable objects:

list
tuple
string
dictionary
The process of going through each item is called iteration or traversal.

```python
## Example for for loop :Resding the list of numbers from a list

num=[0,1,2,3,4]

for i in num:
    print(i)
```

2. Sum of Numbers Using For Loop
Here, we already know the container/list of numbers. So we can use a for loop to read each number and add it to the total.

Important idea:

sum-val = sum-val + number
Short form:

sum_val += number

```python
numbers=[1,2,3,4,5]
sum_val=0

for num in numbers:
   sum_val=sum_val+num
   print("Final sum",sum_val)

print("total sum",sum_val)
```

3. For Loop Using Index Position
   
Python indexing starts from 0.

For example, in this list:

numbers = [1, 2, 3]
index 0 has value 1
index 1 has value 2
index 2 has value 3
We use range(len(list_name)) when we want to work with index positions.


```python
# Sum using index position

numbers = [1, 2, 3, 4, 5]

sum_value = 0

for index in range(len(numbers)):
    print("Index:", index, "Value:", numbers[index])
    sum_value += numbers[index]

print("Final sum:", sum_value)
```

4. For Loop with Strings in a List
If the values are strings, we cannot start with 0 because strings should be added to strings.

So we start with an empty string:
result = ""

```python
# Joining skill words using for loop

fruits = ["Apple", "banana", "grapes", "melon"]

result = ""

for fruit in fruits:
    result += fruit + " "

print(result)
```

```python
# Print String List with Index and Value

#Sometimes we need both the index position and the data at that position.
    
fruits = ["Apple", "banana", "grapes", "melon"]

for index in range(len(fruits)):
    print("Index", index, "has value:", fruits[index])
```

#6. Reverse a String Using For Loop

This is a common interview-style question.

We start with an empty string and keep adding the current character before the existing value.

reversed_text = character + reversed_text


```python
# Reverse a string without using built-in reverse function
text = "Hello"
reversed_text = ""

for index in range(len(text)):
    reversed_text = text[index] + reversed_text

print("Original string:", text)
print("Reversed string:", reversed_text)
```

## While Loop
A while loop is used when we do not have a fixed iterable object, but we know a condition.

The loop continues as long as the condition is True.

Important: In a while loop, we must update the counter or break the condition. Otherwise, it can become an infinite loop.

```python
# Sum of first n natural numbers using while loop

n = 15
sum_value = 0
counter = 1

while counter <= n:
    sum_value += counter
    counter += 1

print("Sum of first", n, "natural numbers is:", sum_value)
```

```python
#8. Countdown Using While Loop:

#If we want to count down, we decrease the counter using:

#count -= 1

# Countdown example

count = 5

while count >= 1:
    print(count)
    count -= 1

print("counting is Done!")
```

Real-Time Use Cases of Loops:
For loop example: In an e-commerce cart, we already have a list of products, so we can use a for loop to calculate the total price.

While loop example: In a login system, the program can keep asking for the correct password until the user enters the correct one.

```python
# For loop real-time example: cart total

prices = [10, 20, 30, 40]

total = 0

for price in prices:
    total += price
    print("Sum value:", total)

print("Total sum value:", total)
    
```

```python
# While loop real-time example without input, so Run All will not stop

correct_password = "Pranitha@325"

attempts = ["Rishi@12", "Prani@12", "Pranitha@123"]

index = 0

while index < len(attempts):

    entered_password = attempts[index]

    print("Entered password is:", entered_password)

    if entered_password == correct_password:
        print("Entered password is correct:", entered_password)
        break

    index += 1
```

```python
# Range examples

for i in range(5):
  print(i)
```

```python
##11. Break Statement

##break is used to stop the loop immediately when a condition is met.

#Example 1: Stop When Number is 3

for num in range(1,6):

        if num == 3:
           break
    
        print(num)       
```

```python
#12. Continue Statement
#continue skips the current iteration and moves to the next iteration.

text = "Bantu"

for character in text:
    if character == "a":
        continue
    print(character)
      
```

```python
#13.String Slicing Practice

City = "Boston"

print("original",City)
print("first two characters",City[0:2])
print("middile two characters",City[2:4])
print("Last two characters",City[-1:-2])
print("from second character to last ",City[1:])
print("Every 2nd character:", City[::2])
print("Reverse string:", City[::-1])
print("Last character:", City[-1])


```
