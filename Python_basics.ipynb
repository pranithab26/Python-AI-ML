## 3. Functions

Functions in Python are used in situations which requires using a specific code block multiple times.

For example, If we need to calculate the area of 10 squares with different side lengths, we can define a function which takes the side length as an argument and returns the area of the square. We can then use ths function directly instead of writing the entire code for calculating area everytime.

```python
def area_of_square(side):
    return side*side
squares = [2,5,6,8,9,10,7,1,3,4]
for i in squares:
    print(area_of_square(i))
```

## 4.Iterators

In Python, an iterator is an object that allows you to traverse (loop through) a collection of data one element at a time without loading all elements into memory at once.

Key Concepts

An iterator must implement two methods:

__iter__() – Returns the iterator object itself.
__next__() – Returns the next value in the sequence. When there are no more items, it raises a StopIteration exception.

```python
numbers = [10, 20, 30, 40]

iterator = iter(numbers)  # Create an iterator

print(next(iterator))  # 10
print(next(iterator))  # 20
print(next(iterator))  # 30
print(next(iterator))  # 40
```
