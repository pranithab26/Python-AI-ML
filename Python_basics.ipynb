## i) Functions

Functions in Python are used in situations when we want to execute specific block of code multiple times then will go for functions.

For example, If we need to calculate the area of 10 squares with different side lengths, we can define a function which takes the side length as an argument and returns the area of the square. We can then use ths function directly instead of writing the entire code for calculating area everytime.

def keyword

```python
def area_of_square(side):
    return side*side
squares = [2,5,6,8,9,10,7,1,3,4]
for i in squares:
    print(area_of_square(i))
```

##  ii)Iterators

An iterator is an object that allows you to traverse (loop through) a collection of data one element at a time.

Key Concepts:

An iterator must implement two methods:

__iter__() – Returns the iterator object itself.
__next__() – Returns the next value in the sequence. When there are no more items, it raises a StopIteration exception.

```python
numbers = [10, 20, 30, 40]

iterator = iter(numbers)  # Create an iterator

print(next(iterator))  # 10
print(next(iterator))  # 20
print(next(iterator))  # 30
print(next(iterator))  # 40
```

## iii) Generators

A generator is a special type of function that allows you to generate values one at a time instead of creating and storing all values in memory at once.

Generators use the yield keyword instead of return.

## Why we use Generators?
-More memory efficient.

-Useful for handling large datasets.

-Values are produced on demand.

```python
def numbers():
    yield 1
    yield 2
    yield 3

gen = numbers()

print(next(gen))
print(next(gen))
print(next(gen))
```

## How it works
- yield 1 pauses the function and returns 1.
  
- When next(gen) is called again, execution resumes from where it stopped.

- This continues until all values are generated.

## iV) Lamda:

A lambda function is a small, anonymous function that can have any number of arguments but only one expression.

It is used when you need a simple function for a short period of time without defining it using def.

## Key Characteristics of Lambda Functions
-Anonymous (no function name required).

-Can take multiple arguments.

-Contains only one expression.

-Automatically returns the result of the expression.

-Commonly used with map(), filter(), and sorted().

## Lambda vs Regular Function
#Regular Function	                          # Lambda Function
Defined using def	                           Defined using lambda
Can have multiple statements	               Only one expression
Has a function name	                           Often anonymous
Better for complex logic	                   Best for simple operations

```python
square = lambda x: x * x

print(square(5))
```

```python
## Adding Two Numbers

add = lambda a, b: a + b

print(add(10, 20))
```

```python
## Finding the Larger Number
maximum = lambda a,b: a if a > b else b
print(maximum(15, 10))
```

```python
## Using lamda with sorted()
emp=[
    ("ravi",5000),
    ("Suresh",7000),
    ("Harish",6000)
]
sorted_emp=sorted(emp,key=lambda emp:emp[1])
print(sorted_emp)
```

```python
## Using lambda with map()
numbers=[1,2,3,4,5]
squares=list(map(lambda x:x*x,numbers))
print(squares)            
```

```python
## Using lambda with filter()
numbers=[1,2,3,4,5]
even=list(filter(lambda x:x%2==0,numbers))
print(even)
```

## filter() 

#The filter() function is used to select items from a list based on a condition.

#It keeps only the values that return True.
# Syntax: 
filter(function, iterable)


```python
## Printing Even numbers
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)

## How it works:
#Checks each number
#Keeps only even numbers
```

```python
## Numbers greater than 3

numbers=[1,2,3,4,5,6]
result=list(filter(lambda x:x>3,numbers))
print(result)
```

```python
##reduce() :The reduce() function is used to combine all elements into a single result.

##It applies a function again and again (cumulatively).


## Example:Sum of the numbers

from functools import reduce

numbers = [1, 2, 3, 4, 5]

result = reduce(lambda x, y: x + y, numbers)

print(result)

##How it works:

1 + 2 = 3
3 + 3 = 6
6 + 4 = 10
10 + 5 = 15

```

```python
## print() and return in Python
##print() is used to display output on the screen.

##print() is used to display information on the console, whereas return is used to send a value back from a function. A value returned by return can be stored in a variable and used later in the program, while print() only displays the value and does not make it available for further processing.

def add(a, b):
    print(a + b)

add(5, 3)
```

```python
##return Statement

##return is used to send a value back from a function.

def add(a, b):
    return a + b

result = add(6, 5)

print(result)
```

```python
##Key Difference
## Using print()

def square(x):
    return x * x

result = square(5)

print(result)

##The value is returned and can be reused later.

```

```python
##Reusing a Returned Value

def square(x):
    return x * x

result = square(5)

print(result + 10)

##This works because return gives the value back to the program.
```

| `print()`                                              | `return`                                       |
| ------------------------------------------------------ | ---------------------------------------------- |
| Displays output on the screen                          | Sends a value back from a function             |
| Used for debugging or showing information              | Used when the result is needed later           |
| Does not make the value available outside the function | Makes the value available outside the function |
| Function continues after `print()`                     | Function ends when `return` is executed        |
| Returns `None` by default                              | Returns the specified value                    |
