```python
## Lists, Tuples, Sets, Dictionaries, and Functions

#This assignment includes simple and working solutions.

#Task 1: Create a list of Hobbies and print each one by one.

Hobbies_list = ["singing","Dancing","Playing","Painting"]

for Hobbies in Hobbies_list:
    print("list of Hobbies:",Hobbies)
```

```python
## Task 2: Append, remove, reverse, and sort list values

Hobbies_list = ["singing","Dancing","Playing","Painting"]

Hobbies_list.append("swimming")
Hobbies_list.remove("singing")
Hobbies_list.reverse()
print("After reverse:",Hobbies_list)
Hobbies_list.sort()
print("After sort:",Hobbies_list)    
```

```python
#Task 3: Create a nested list for semesters and access values

term_1 = ["Math", "Physics", "Chemistry"]
term_2 = ["Biology", "History", "Geography"]
term_3 = ["Economics", "Programming", "Statistics"]

terms = [term_1, term_2, term_3]

print("All terms:", terms)
print("Term 2:", terms[1])
print("Programming location:", terms[2][1])
```

```python
##Task 4: Use list comprehension to filter scores

marks = [45, 78, 82, 60, 90, 67, 99]
high_marks = [mark for mark in marks if mark >= 80]

print("All marks:", marks)
print("High marks:", high_marks)
```

```python
## Task 5: Create a tuple and show immutability safely
colors = ("Red", "Green", "Blue")
print("Tuple:", colors)

try:
    colors[1] = "Yellow"
except TypeError as error:
    print("Cannot modify tuple directly:", error)
```

```python
## Task 6: Convert tuple to list, update, and convert back

employee_info = ("John Smith", "TechCorp", "Developer")

temp_list = list(employee_info)
temp_list[1] = "Google"
employee_info = tuple(temp_list)

print("Updated tuple:", employee_info)
```

```python
##Task 7: Use set to remove duplicate skills

languages = ["Java", "C++", "Java", "Python", "C++", "Go"]

unique_languages = set(languages)

print("Original languages:", languages)
print("Unique languages:", unique_languages)
```

```python
## Task 8: Create dictionary profile and read values using keys

profile = {
    "name": "Smith",
    "company": "TechCorp",
    "role": "Software Engineer",
    "skills": ["Java", "AWS", "Docker"]
}

print("Name:", profile["name"])
print("Company:", profile["company"])
print("First skill:", profile["skills"][0])
```

```python
##Task 9: Create nested dictionary for profile

profile = {
    "personal": {"name": "Smith", "city": "Austin"},
    "job": {"company": "TechCorp", "role": "Software Engineer"},
    "technical": {"skills": ["Java", "AWS", "Docker"]}
}

print("Name:", profile["personal"]["name"])
print("Role:", profile["job"]["role"])
print("Second skill:", profile["technical"]["skills"][1])
```

```python
## Task 10: Use dictionary comprehension

numbers = range(1, 11)
squares = {number: number ** 2 for number in numbers}

print("Squares dictionary:", squares)
```

```python
## Task 11: Create a function to summarize my profile

def create_employee_summary(name, company, skills):
    unique_skills = list(set(skills))
    profile = {
        "name": name,
        "company": company,
        "unique_skills": unique_skills,
        "skill_count": len(unique_skills)
    }
    return profile

summary = create_employee_summary(
    "John Smith",
    "TechCorp",
    ["Java", "AWS", "Java", "Docker", "AWS"]
)

print(summary)
```
